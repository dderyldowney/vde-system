```python
# Pause and Resume Execution with Saved State

This notebook demonstrates how to save computation state to disk and resume later by loading that state.
</VSCode.Cell>

<VSCode.Cell language="python">
# Import Required Libraries
import pickle
import os

print("Libraries imported")
</VSCode.Cell>

<VSCode.Cell language="markdown">
## Create and Save State

We'll define a simple state (e.g., a counter) and serialize it to a file.
</VSCode.Cell>

<VSCode.Cell language="python">
state = {'counter': 0, 'data': []}

# save state
with open('state.pkl', 'wb') as f:
    pickle.dump(state, f)

print('State saved:', state)
</VSCode.Cell>

<VSCode.Cell language="markdown">
## Simulate a Pause

Imagine the computation stops here (kernel shutdown or notebook closed). We'll just note it.
</VSCode.Cell>

<VSCode.Cell language="python">
print('--- execution paused ---')
</VSCode.Cell>

<VSCode.Cell language="markdown">
## Load State and Resume Execution

Now we'll load the saved state and continue incrementing the counter.
</VSCode.Cell>

<VSCode.Cell language="python">
# load state
with open('state.pkl', 'rb') as f:
    state = pickle.load(f)

print('Loaded state:', state)

# resume work
for i in range(5):
    state['counter'] += 1
    state['data'].append(state['counter'] * 10)

print('Updated state:', state)

# save again
with open('state.pkl', 'wb') as f:
    pickle.dump(state, f)
</VSCode.Cell>

<VSCode.Cell language="markdown">
## Handle Exceptions During Resume

Wrap loading in try/except to handle missing or corrupt files.
</VSCode.Cell>

<VSCode.Cell language="python">
try:
    with open('state.pkl', 'rb') as f:
        state = pickle.load(f)
except (FileNotFoundError, pickle.UnpicklingError) as e:
    print('Error loading state:', e)
    state = {'counter': 0, 'data': []}

print('Recovered state:', state)
```
